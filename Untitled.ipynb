### Prepre the dataset

```python
# independent and dependent features
X=dataset.iloc[:,:-1]
y=dataset.iloc[:,-1]
```

```python
# train test split
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=123)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
X_test
```

```python
y_test
```

```python
# stadadizing the values
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
# why do we apply stadardsclaer?
# >internally we use the gradient descent our main aim to comes to global minima we have make sure that our indeoendent feature vaues should be the same unit. 
```

```python
from sklearn.linear_model import LinearRegression
lg=LinearRegression()
lg.fit(X_train,y_train)
```

```python
# print the coffeficients and the intercepts
print(lg.coef_)
print(lg.intercept_)
```

```python
# on which parameters the model has been trained 
lg.get_params()
```

```python
# prediction with the test data
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
reg_pred=lg.predict(X_test)
```

```python
print(r2_score(reg_pred,y_test))
```

```python
# plot for predictions
plt.scatter(reg_pred,y_test)
```

```python
#residuals
residuals=y_test-reg_pred
```

```python
residuals
```

```python
# plot the residuals
```

```python
sns.displot(residuals,kind='kde')
```

```python
# sclatter plot with respect to predictions
#uniform distibution
plt.scatter (reg_pred,residuals)
```

```python
print(mean_squared_error(reg_pred,y_test))
```

```python
print(np.sqrt(mean_squared_error(reg_pred,y_test)))
```

```python
print(mean_absolute_error(reg_pred,y_test))
```
